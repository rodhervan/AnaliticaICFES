```python
# Importaciones necesarias
import pandas as pd
import mlflow
import mlflow.sklearn

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import classification_report, accuracy_score
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer
```

```python
# Cargar datos
df = pd.read_csv("df_encoded.csv")

# Datos icfes encoded
X = df.drop(['puntaje_cat'], axis=1)
y = df["puntaje_cat"]

# Dividir datos
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
# Configurar MLflow
mlflow.set_tracking_uri("http://localhost:5000")
experiment = mlflow.set_experiment("xgboost-clas-icfes")

# Definir hiperparámetros para ensayar
param_grid = [
    {"n_estimators": 150, "learning_rate": 0.25, "max_depth": 8},
    {"n_estimators": 150, "learning_rate": 0.2, "max_depth": 8},
    {"n_estimators": 150, "learning_rate": 0.15, "max_depth": 8}
]

for params in param_grid:
    with mlflow.start_run(experiment_id=experiment.experiment_id):
        # Crear y entrenar el modelo
        model = XGBClassifier(
            n_estimators=params["n_estimators"],
            learning_rate=params["learning_rate"],
            max_depth=params["max_depth"],
            use_label_encoder=False,
            eval_metric="mlogloss",
            random_state=42
        )
        model.fit(X_train, y_train)

        # Predicciones
        y_pred = model.predict(X_test)

        # Métricas
        acc = accuracy_score(y_test, y_pred)
        report = classification_report(y_test, y_pred, output_dict=True)

        # Registrar parámetros
        mlflow.log_params(params)
        mlflow.log_metric("accuracy", acc)
        mlflow.log_metric("precision_macro", report["macro avg"]["precision"])
        mlflow.log_metric("recall_macro", report["macro avg"]["recall"])
        mlflow.log_metric("f1_macro", report["macro avg"]["f1-score"])

        # Log del modelo
        mlflow.sklearn.log_model(model, "xgb-model")

        # Guardar localmente
        #joblib.dump(model, f"xgb_model_{params['n_estimators']}_{params['learning_rate']}.pkl")

        print(f"Ensayo completado con accuracy: {acc:.4f} - Hiperparámetros: {params}")
```
